```python
import numpy as np
from numpy import random
import matplotlib.pyplot as plt

arr_3 = np.array([110, 20, 32, 24])
arr_4 = np.array([12, 17, 67, 80])

print (arr_3 * arr_4)
print (arr_3 + arr_4)
print (arr_3 - arr_4)
print (arr_3 / arr_4)


arr_5 = np.random.randint (3, 100, size= (4, 6))
print (arr_5)
print ()
print (f"The minimum number is: {np.min (arr_5)}")
print (f"The maximum number is: {np.max (arr_5)}")
avg = ((np.max (arr_5) + np.min (arr_5))/2 )
print (f"The average number is: {avg}")
```
